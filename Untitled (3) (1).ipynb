```python
# 1. Imports and Setup
# ----------------------------
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.preprocessing import image_dataset_from_directory
from sklearn.metrics import confusion_matrix, classification_report

# Set random seed for reproducibility
seed = 42
tf.random.set_seed(seed)
np.random.seed(seed)
```

```python
# 2. Dataset Directory Configuration

# Update the base directory if different.
BASE_DIR = "IP102dataset"
train_dir = os.path.join(BASE_DIR, "train")
val_dir   = os.path.join(BASE_DIR, "val")
test_dir  = os.path.join(BASE_DIR, "test")

# Parameters for loading images
IMG_SIZE = (224, 224)  # resize images to 224x224
BATCH_SIZE = 32
```

```python
# 3. Loading the Dataset
# ----------------------------
# Using image_dataset_from_directory to load data and infer classes from folder names.
train_ds = image_dataset_from_directory(
    train_dir,
    labels="inferred",
    label_mode="categorical",  # returns one-hot encoded labels
    batch_size=BATCH_SIZE,
    image_size=IMG_SIZE,
    shuffle=True,
    seed=seed
)

val_ds = image_dataset_from_directory(
    val_dir,
    labels="inferred",
    label_mode="categorical",
    batch_size=BATCH_SIZE,
    image_size=IMG_SIZE,
    shuffle=True,
    seed=seed
)

test_ds = image_dataset_from_directory(
    test_dir,
    labels="inferred",
    label_mode="categorical",
    batch_size=BATCH_SIZE,
    image_size=IMG_SIZE,
    shuffle=False
)

# Fetch the class names from the training dataset
class_names = train_ds.class_names
print("Number of classes:", len(class_names))
print("Classes:", class_names)
```

```python
# 4. Visualizing Sample Images from Training Data
# ----------------------------
plt.figure(figsize=(10, 10))
for images, labels in train_ds.take(1):
    for i in range(9):
        ax = plt.subplot(3, 3, i + 1)
        # Convert tensor values into uint8 image format for display
        plt.imshow(images[i].numpy().astype("uint8"))
        # Get the index of the maximum value (one-hot vector) for the label
        label_index = np.argmax(labels[i])
        plt.title(class_names[label_index])
        plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# 5. Data Augmentation and Optimization Setup
# ----------------------------
# Create a small data augmentation pipeline. This will be applied to the inputs during training.
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1)
])
# For performance, configure dataset prefetching
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=AUTOTUNE)
test_ds = test_ds.prefetch(buffer_size=AUTOTUNE)
```

```python
import ssl
ssl._create_default_https_context = ssl._create_unverified_context

import tensorflow as tf
from tensorflow.keras import layers, models


# 6. Building the Model (Transfer Learning with MobileNetV2)
# ----------------------------
# Load the base model pre-trained on ImageNet. Do not include the top (classifier) layers.
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3),
    include_top=False,
    weights="imagenet"
)
base_model.trainable = False  # freeze the base model weights

# Create the full model
inputs = tf.keras.Input(shape=(IMG_SIZE[0], IMG_SIZE[1], 3))
x = data_augmentation(inputs)  # apply augmentation
x = tf.keras.applications.mobilenet_v2.preprocess_input(x)  # apply MobileNetV2-specific preprocessing
x = base_model(x, training=False)  # pass through base model
x = layers.GlobalAveragePooling2D()(x)  # global pooling
x = layers.Dropout(0.2)(x)  # dropout for regularization
outputs = layers.Dense(len(class_names), activation="softmax")(x)  # classifier layer
model = tf.keras.Model(inputs, outputs)

# Compile the model with optimizer, loss, and metrics
model.compile(optimizer=optimizers.Adam(learning_rate=0.001),
              loss="categorical_crossentropy",
              metrics=["accuracy"])

# Print a model summary
model.summary()
```

```python
# 7. Training the Model
# ----------------------------
# Train the model using the training and validation datasets
EPOCHS = 5
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS
)
```

```python
# 8. Visualizing Training Metrics
# ----------------------------
# Plot training & validation accuracy and loss at each epoch.
acc = history.history["accuracy"]
val_acc = history.history["val_accuracy"]
loss = history.history["loss"]
val_loss = history.history["val_loss"]
epochs_range = range(EPOCHS)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label="Training Accuracy")
plt.plot(epochs_range, val_acc, label="Validation Accuracy")
plt.legend(loc="lower right")
plt.title("Training and Validation Accuracy")

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label="Training Loss")
plt.plot(epochs_range, val_loss, label="Validation Loss")
plt.legend(loc="upper right")
plt.title("Training and Validation Loss")
plt.show()
```

```python
# 9. Evaluating on the Test Set
# ----------------------------
test_loss, test_accuracy = model.evaluate(test_ds)
print("Test Loss: {:.4f}".format(test_loss))
print("Test Accuracy: {:.4f}".format(test_accuracy))
```

```python
# 10. Making Predictions and Generating Evaluation Visualizations
# ----------------------------
# Collect the true labels and predictions from the test dataset
y_true = []
y_pred = []

# Iterate over the test dataset batches.
for images, labels in test_ds:
    preds = model.predict(images)
    y_true.extend(np.argmax(labels, axis=1))
    y_pred.extend(np.argmax(preds, axis=1))

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(20, 20))
sns.heatmap(cm, annot=False, cmap="Blues", xticklabels=class_names, yticklabels=class_names)
plt.title("Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()

# Classification Report
print("Classification Report:\n")
print(classification_report(y_true, y_pred, target_names=class_names))
```

```python
# 11. Saving the Model
# ----------------------------
# Save the model in HDF5 format for later use.
model.save("ip02_pest_model.h5")
print("Model saved as ip02_pest_model.h5")
```
